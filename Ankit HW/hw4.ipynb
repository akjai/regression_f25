```python
import numpy as np
import pandas as pd
from statsmodels.formula.api import ols
from plotnine import ggplot, aes, geom_point, labs
from scipy.stats import linregress
import statsmodels.api as sm
```

```python
df = pd.read_csv("../data/KelleyBlueBookData.csv")
df.head(5)
```

```python
# For the dataset KelleyBlueBookData.csv, response= Price against the following predictors: Mileage, Liter, 
# Cylinder (in this order). Treat Cylinder as a quantitative variable.
model = ols('Price ~ Mileage + Liter + Cylinder', data=df).fit()
model.summary()
```

# 4a. 

```python
sm.stats.anova_lm(model, typ=1) #sequential
```

The null hypothesis is that the coefficient for cylinder is 0, meaning that cylinder is not significant to the model. The alternate hypothesis is that the coefficient is not zero, meaning cylinder is significant. For including the cylinder predictor, the F-stat is 15.997, p-value is <0.001. For excluding the cylinder predictor, the F-stat is 375.436, p-value is <0.001. We reject the null hypothesis because our p-value is <0.001, meaning cylinder is a significant predictor for price.


# 4b. 

```python
y = df['Price']
y_hat = model.predict(df)
mean = df['Price'].mean()
n = len(y)
p = 3

sst = ((y - mean)**2).sum()
sse = ((y - y_hat)**2).sum()

r2 = 1 - (sse/sst)
r2_adjusted = 1 - ((sse / (n - p)) / (sst / (n - 1)))

print(r2)
r2_adjusted
```

Yes, they match.

The F-test does not match from 4a because it tests each predictor after factoring in all other predictors instead of sequentially, so the overall result will be different.
